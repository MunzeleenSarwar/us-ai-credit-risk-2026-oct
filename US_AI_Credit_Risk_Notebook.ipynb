# Credit Risk Analysis: 10 AI Companies

This notebook recomputes the latest credit metrics from data/metrics.csv. Run the download script first, or use the saved files in data/.

```
import pandas as pd
m=pd.read_csv('data/metrics.csv')
m=m.sort_values('fiscal_year_end')
lat=m.groupby('ticker').tail(1)
print(lat[['ticker','company','fiscal_year_end','total_debt','debt_ebitda','interest_coverage','current_ratio','cash_to_debt']].to_string(index=False))
```

```
# Stress test: earnings fall 30%
lat=lat.copy()
lat['coverage_stress']=lat.interest_coverage*0.7
lat['debt_ebitda_stress']=lat.debt_ebitda/0.7
print(lat[['ticker','interest_coverage','coverage_stress','debt_ebitda','debt_ebitda_stress']].to_string(index=False))
```
